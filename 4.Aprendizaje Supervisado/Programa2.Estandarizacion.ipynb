```python
from sklearn import preprocessing
import numpy as np
```

```python
# Carga de datos.
X_train = np.array([[ 1., -1.,  2.],
                    [ 2.,  0.,  0.],
                    [ 0.,  1., -1.]])
```

```python
# Limpieza de datos: estandarización.
standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_norm = standardizer.fit_transform(X_train)
print(X_train_norm)
```

```python
X_train_2 = np.array([[ 1., -1.,  2.],
                    [ 2.,  0.,  0.],
                    [ 0.,  1., -1.]])
```

```python
# Aplicamos los métodos "fit" y "transform" por separado
X_train_norm2 = standardizer.fit(X_train_2).transform(X_train_2)
print(X_train_norm2)
```

```python
# Comprobar si las matrices estandarizadas son iguales
X_train_norm == X_train_norm2
```
